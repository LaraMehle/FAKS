```python
import numpy as np
import matplotlib.pyplot as plt

# Signal (c) — stopničast signal
I = np.array([1, 1, 0, 0, 0.3, 0.5, 0.5, 0, 0, 1])

# Kernel (a) — Gaussov (glajenje)
def gauss(sigma):
    radius = int(np.ceil(3 * sigma))
    x = np.arange(-radius, radius + 1)
    g = (1 / (np.sqrt(2 * np.pi) * sigma)) * np.exp(-(x**2) / (2 * sigma**2))
    g /= g.sum()
    return g

k1 = gauss(1)

# Kernel (b) — valovit (robni detektor)
k2 = np.array([-1, 0, 1])  # preprosta verzija robnega detektorja

# Konvolucije
conv1 = np.convolve(I, k1, mode='same')
conv2 = np.convolve(I, k2, mode='same')

# Risanje
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.plot(I, marker='o')
plt.title('Signal (c)')

plt.subplot(1, 3, 2)
plt.plot(conv1, marker='o', color='blue')
plt.title('Gaussov kernel * signal (glajenje)')

plt.subplot(1, 3, 3)
plt.plot(conv2, marker='o', color='red')
plt.title('Valovit kernel * signal (robni odziv)')

plt.tight_layout()
plt.show()
```
